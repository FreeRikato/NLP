### 1. Write a program to remove punctuations, special symbols, numbers using regular expression

- **Regex Pattern**: `r'[^\w\s]|\d'` removes anything that is not a word character or whitespace, and also removes digits.
- **Function Purpose**: The function `remove_punct_symbols_numbers` cleans the text by stripping out punctuation, special symbols, and numbers, which aligns with the experiment goal.
- **Output**: Demonstrates successful removal of punctuations, symbols (like "@", ".", etc.), and numbers (e.g., "123").

```python
import re
def remove_punct_symbols_numbers(text):
    return re.sub(r'[^\w\s]|\d', '', text)

print("\nText after removing punctuations, special symbols, and numbers:")
cleaned_text = remove_punct_symbols_numbers(sample_text)
print(cleaned_text)
```

### 3. Perform stemming and lemmatization on text

- **Tokenization**: Splits the input text into individual words (tokens) to apply stemming and lemmatization.
- **Stemming**: Reduces words to their root forms (e.g., "jumping" to "jump") using the PorterStemmer.
- **Lemmatization**: Converts words to their dictionary form (e.g., "foxes" to "fox") using WordNetLemmatizer.
- **Relevance**: The code appropriately applies both techniques to demonstrate the difference between stemming (more aggressive reduction) and lemmatization (context-aware word form reduction).

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

def process_text(text):
    tokens = word_tokenize(text)
    return {
        'original': tokens,
        'stemmed': [PorterStemmer().stem(word) for word in tokens],
        'lemmatized': [WordNetLemmatizer().lemmatize(word) for word in tokens]
    }

# Example usage
text = "The quick brown foxes are jumping over the lazy dogs"
result = process_text(text)

for key, value in result.items():
    print(f"{key.capitalize()}:", value)
```

### 4. Find the most common words in the text to exclude as stopwords


- **Regex filtering**: It breaks the text into words, converting them to lowercase for uniformity.
- **Word frequency**: Uses `Counter` to count word occurrences and returns the `n` most common.
- **Relevance**: The experiment aims to find common words to exclude as stopwords, which this code directly achieves by outputting the most frequent words.

```python
from collections import Counter
import re

def find_common_words(text, n=10):
    words = re.findall(r'\w+', text.lower())
    return [word for word, _ in Counter(words).most_common(n)]

# Example usage
text = "Your long text goes here. This is just a sample text to demonstrate the function."
print(find_common_words(text))
```

### 6. POS Tagger (Plain)

- **POS tagging**: The code uses SpaCy's `en_core_web_sm` model to perform part-of-speech (POS) tagging on the input text, returning a list of tokens with their corresponding POS tags.
- **Visualization**: The `displacy.render` function visualizes the dependency parse tree of the text with options for a compact display and color customization (green).

```python
!python -m spacy download en_core_web_sm
```

```python
import spacy
from spacy import displacy

def pos_tag_and_visualize(text):
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)

    # POS tagging
    pos_tags = [(token.text, token.pos_) for token in doc]

    # Visualization
    html = displacy.render(doc, style="dep", options={"compact": True, "color": "blue"})

    return pos_tags, html

# Example usage
text = "The quick brown fox jumps over the lazy dog."
tags, viz_html = pos_tag_and_visualize(text)

print("POS Tags:", tags)
print("\nVisualization HTML:")
print(viz_html)
```

### 7. Syntax Parsing Using CFG

- **CFG Grammar**: Defines sentence structure with specific rules (S -> NP VP, etc.).
- **Parse Tree**: Visual output shows how the sentence is broken into components (NP, VP, etc.).
- **ChartParser**: Efficiently parses the sentence based on the grammar.

```python
import nltk
from nltk import CFG

# Define a simple CFG grammar
grammar = CFG.fromstring("""
    S -> NP VP
    NP -> DT NN
    VP -> VBZ NP
    DT -> 'the'
    NN -> 'cat' | 'mat'
    VBZ -> 'sits'
""")

# Input sentence to parse
sentence = ['the', 'cat', 'sits', 'the', 'mat']

# Create a parser
parser = nltk.ChartParser(grammar)

# Parse the sentence
for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

### 8. Named Entity Recognition


- `perform_ner(text)` function uses `spaCy` to process the input text and identify entities.
- It correctly identifies entities like "Apple" (ORG), "U.K." (GPE), and "$1 billion" (MONEY), aligning with the NER task of classifying entities into categories.

```python
import spacy
from spacy import displacy

def perform_ner(text):
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)
    return doc

# Example usage
text = "Apple is looking at buying U.K. startup for $1 billion"
doc = perform_ner(text)

# Print entities
for ent in doc.ents:
    print(f"{ent.text}: {ent.label_}")

# Generate HTML visualization
displacy.render(doc, style="ent")
```

### 9. Word Sense Disambiguation

- Identifies the target word in the provided context.
- Visualizes the sentence structure with dependency parsing using `displacy`.
- Supports comparing different contexts of ambiguous words like "bank."

```python
import spacy
from spacy import displacy

def disambiguate_and_visualize(word, context):
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(context)
    target = next(token for token in doc if token.text.lower() == word.lower())

    # Generate HTML visualization
    displacy.render(doc, style="dep", options={"compact": True, "color": "blue"})

# Example usage
word = "bank"
context1 = "I went to the bank to deposit money."
context2 = "The river bank was covered in flowers."

disambiguate_and_visualize(word, context1)
disambiguate_and_visualize(word, context2)
```

### 10. Brill Tagger

- **Loads data**: Uses 3000 sentences from the Treebank corpus for training and the remainder for testing.
- **Unigram tagger**: Serves as the initial tagger for the Brill transformation-based tagger.
- **Brill templates**: Defines transformation rules using a pre-defined template (`fntbl37()`).
- **Brill trainer**: Trains the Brill tagger, learning up to 100 transformation rules.
- **Evaluation**: Evaluates the tagger on the test set with an accuracy of 88.50%.
- **Tagging**: Demonstrates tagging of a sample sentence but shows errors (tags missing) for unknown words.

```python
nltk.download('treebank')
nltk.download('universal_tagset')
```

```python
import nltk
from nltk.corpus import treebank
from nltk.tag import brill, brill_trainer

# Load data
train = treebank.tagged_sents(tagset='universal')[:3000]
test = treebank.tagged_sents(tagset='universal')[3000:]

# Initialize Unigram Tagger and Brill templates
ut = nltk.UnigramTagger(train)
templates = brill.fntbl37()

# Train Brill Tagger
trainer = brill_trainer.BrillTaggerTrainer(initial_tagger=ut, templates=templates, deterministic=True)
bt = trainer.train(train, max_rules=100)

# Evaluate and use
print(f"Accuracy: {bt.evaluate(test):.4f}")
print(bt.tag("The quick brown fox jumps over the lazy dog".split()))
```

### 12. BOW with TF-IDF

- **BoW and TF-IDF:** Key techniques for text representation in NLP.
- **BoW (CountVectorizer):** Converts text into a matrix of token counts (term frequency).
- **TF-IDF (TfidfVectorizer):** Converts text into a matrix of weighted values reflecting term importance across the corpus.
- **Vocabulary consistency:** Same vocabulary used for BoW and TF-IDF, ensuring comparison.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Sample text data
corpus = [
    'I love machine learning',
    'NLP is fascinating',
    'I love coding and machine learning',
    'NLP is part of AI'
]

# Create a CountVectorizer for Bag of Words (BoW)
bow_vectorizer = CountVectorizer()
bow = bow_vectorizer.fit_transform(corpus)

# Create a TfidfVectorizer for TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf = tfidf_vectorizer.fit_transform(corpus)

# Display BoW and TF-IDF results
print("Bag of Words:\n", bow.toarray())
print("BoW Vocabulary:\n", bow_vectorizer.get_feature_names_out())
print("\nTF-IDF:\n", tfidf.toarray())
print("TF-IDF Vocabulary:\n", tfidf_vectorizer.get_feature_names_out())
```

### 13. N‐Gram Language Model

- **Tokenization**: The corpus is tokenized into words, essential for creating n-grams.
- **N-grams creation**: Code generates n-grams (unigrams here as n=1) from the tokens.
- **Frequency count**: Uses `Counter` to calculate and output n-gram frequencies.
- **Corpus analysis**: The code shows how frequently each unigram (single word) appears in the corpus.
- **Flexibility**: Allows easy adjustment of `n` to analyze bigrams, trigrams, etc.

```python
from nltk import ngrams
from collections import Counter
import nltk

# Sample corpus
corpus = "I love natural language processing. NLP is fascinating. I enjoy learning NLP."

# Tokenize the corpus into words
tokens = nltk.word_tokenize(corpus.lower())

# Create n-grams (n can be 2 for bigrams, 3 for trigrams, etc.)
n = 1
n_grams = list(ngrams(tokens, n))

# Count frequency of n-grams
n_gram_freq = Counter(n_grams)

# Output the n-grams and their frequencies
print(n_gram_freq)
```

### 14. Word2Vec

- **Word2Vec** model is trained on a tokenized corpus using Gensim.
- **Corpus** is a list of sentences broken into individual words.
- **Model Parameters**: `vector_size=10`, `window=2` defines vector dimensions and context window size.
- **Vocabulary** shows words learned by the model after training.
- **Word Embedding**: Prints the vector for the word "learning".
- **Similarity**: Finds the most similar words to "learning" based on vector distance.

```python
!pip -q install gensim
```

```python
# Import the library
from gensim.models import Word2Vec

# Define your corpus (a list of tokenized sentences)
sentences = [['i', 'love', 'machine', 'learning'],
             ['deep', 'learning', 'is', 'fun'],
             ['i', 'enjoy', 'nlp', 'tasks']]

# Train the Word2Vec model
model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, workers=4)

# Save the word vectors and vocab
words = list(model.wv.index_to_key)

# Print the vocabulary
print(f"Vocabulary: {words}")

# Find the vector for a specific word
print(f"Vector for 'learning': {model.wv['learning']}")

# Find most similar words
print(f"Most similar words to 'learning': {model.wv.most_similar('learning')}")
```

### 15. GloVe

- **Glove model**: Code implements the GloVe algorithm, which is designed to generate word embeddings based on co-occurrence data.
- **Corpus creation**: The code defines a small sample text corpus for the experiment, which is processed into a co-occurrence matrix using the `Corpus` class.
- **Window size**: Uses a window size of 10, controlling how many words are considered as context for each target word.
- **Model training**: The GloVe model is trained with 100 components (dimensionality of embeddings) over 20 epochs.
- **Learning rate**: A learning rate of 0.05 ensures gradual weight updates during training.
- **Parallelism**: The code uses 4 threads to speed up the model training.
- **Word similarity**: The output shows similarity scores for words in the corpus, demonstrating that embeddings have been successfully trained.

```python
!pip install glove-python3
```

```python
from glove import Corpus, Glove

# Sample text corpus
corpus = [['this', 'is', 'a', 'sample'], ['glove', 'for', 'word', 'embeddings']]

# Create the co-occurrence matrix
corpus_model = Corpus()
corpus_model.fit(corpus, window=10)

# Train the GloVe model
glove = Glove(no_components=100, learning_rate=0.05)
glove.fit(corpus_model.matrix, epochs=20, no_threads=4, verbose=False)
glove.add_dictionary(corpus_model.dictionary)

# Find words similar to 'glove'
print(glove.most_similar('glove'))
```

### 16. Fast Text

- **Gensim's FastText model** is used for training word embeddings on a sample set of sentences.
- **Sentences are tokenized** to prepare them for input into the model.
- **FastText captures subword information**, making it useful for understanding words in varied forms or uncommon words.
- **Model configuration**: 10-dimensional vectors, a context window of size 2, skip-gram model disabled (cbow used), and all words considered (min_count=1).
- **Word vector output**: The code retrieves and prints the learned vector for the word "this", illustrating how the model converts words into vectors based on their context.

```python
!pip -q install gensim
```

```python
from gensim.models import FastText

# Sample sentences for FastText training
sentences = [
    'This is the first sentence.',
    'This is the second sentence.',
    'And here is the third one.',
    'Finally, this is the fourth sentence.'
]

# Tokenize the sentences
tokenized_sentences = [sentence.lower().split() for sentence in sentences]

# Create the FastText model
model = FastText(sentences=tokenized_sentences, vector_size=10, window=2, min_count=1, sg=0)

# Get the vector for a specific word
vector = model.wv['this']

# Display the vector
print('Vector for the word "this":', vector)
```

### 17. Spam Email Detection

- Uses a pre-trained model from Hugging Face for text classification, specific to spam email detection.
- The `pipeline` function simplifies model loading and usage, fitting the experiment's need for classifying emails as spam or not.
- The `classify_email` function processes email content and determines if it is spam using the model's prediction.
- The experiment outputs either "Spam" or "Not Spam," aligning with the task of detecting spam in email content.
- The provided example correctly identifies spam, verifying the code's appropriateness for the experiment.

```python
!pip -q install transformers
```

```python
from transformers import pipeline

# Load the pre-trained spam detection model from Hugging Face
spam_classifier = pipeline("text-classification", model="nathan-marquez/distilbert-spamEmail-base")

# Function to classify email content as spam or not
def classify_email(email_content):
    result = spam_classifier(email_content)
    label = result[0]['label']
    return "Spam" if label == "LABEL_1" else "Not Spam"

# Example usage
email_content = "Congratulations! You've won a free lottery. Click here to claim your prize!"
print(classify_email(email_content))
```

### 18. Context-Aware Sentiment Analysis (Hate Speech Detection)

- **Pre-trained model**: Uses `bert-base-uncased-emotion` for emotion classification, appropriate for detecting sentiment, a core part of hate speech detection.
- **Sentiment classification**: Labels email texts with emotions (joy, anger), enabling context-aware sentiment analysis.
- **Hate speech relevance**: The model identifies negative emotions (like anger), which are relevant indicators for hate speech detection.
- **Example texts**: Includes both benign and emotionally charged examples (e.g., "I hate this thing") for varied sentiment analysis.
- **Predictions**: Outputs a confidence score for each emotion, useful for gauging model certainty in detecting potential hate speech.

```python
!pip -q install transformers
```

```python
from transformers import pipeline

# Load a pre-trained model for text classification
classifier = pipeline('text-classification', model="bhadresh-savani/bert-base-uncased-emotion")

# Example email texts
email_texts = [
    "Congratulations! You've won a free iPhone. Click here to claim your prize.",
    "Hi, just wanted to check in and confirm our meeting tomorrow.",
    "Exclusive offer! Buy one get one free on all electronics.",
    "I hate this thing."
]

# Predict whether the emails are spam or not
predictions = classifier(email_texts)

# Display results
for text, pred in zip(email_texts, predictions):
    print(f"Text: {text}\nPrediction: {pred}\n")
```
